## 7. Compliance, Ethics & Disclaimer

The system separates predictive output from investigative evidence and keeps synthetic ground truth in a validation-only role.

> **This is an auditable research-grade proof of concept using synthetic data. It is not a production AML monitoring platform, regulator-approved model, autonomous financial-crime decision system, or replacement for trained investigators.**
